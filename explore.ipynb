```python
def cosine_similarity(vec1, vec2):
    dot_product = sum(a * b for a, b in zip(vec1, vec2))
    magnitude1 = sum(a * a for a in vec1) ** 0.5
    magnitude2 = sum(b * b for b in vec2) ** 0.5
    if magnitude1 == 0 or magnitude2 == 0:
        return 0
    return dot_product / (magnitude1 * magnitude2)

item1_asin = "B084KJLXPB"
item2_asin = "B08SCCH6NC"

for i in range(80):
    data = pd.read_parquet(f"{FOLDER_NAME}/{EMBEDDING_MODEL_NAME}/text_embeddings_Beauty_and_Personal_Care_{i:05d}.parquet")
    if item1_asin in data["asin"].values:
        item1_embedding = data.loc[data["asin"] == item1_asin, "text_embedding"].values[0]
    if item2_asin in data["asin"].values:
        item2_embedding = data.loc[data["asin"] == item2_asin, "text_embedding"].values[0]

display(cosine_similarity(item1_embedding, item2_embedding))
```
