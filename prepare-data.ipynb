# Prepare Data

```python
from datasets import load_dataset
```

```python
def load_reviews(category: str):
    dataset = load_dataset("McAuley-Lab/Amazon-Reviews-2023", f"raw_review_{category}")
    return dataset["full"]

def load_items(category: str):
    dataset = load_dataset("McAuley-Lab/Amazon-Reviews-2023", f"raw_meta_{category}")
    return dataset["full"]
```

```python
load_reviews('All_Beauty').to_pandas()
```
